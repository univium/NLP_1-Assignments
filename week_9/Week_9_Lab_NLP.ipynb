#Sequential Model Keras

- There are two ways to build Keras models: sequential and functional. 
- The sequential API allows you to create models layer-by-layer for most problems. 

- More here: https://machinelearningmastery.com/keras-functional-api-deep-learning/

```python
#libraries
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
```

```python
#Some useful built in RNN layers in Keras
#check out: https://www.tensorflow.org/api_docs/python/tf/keras/layers/SimpleRNN

#define a sequential model
model = keras.Sequential()

# Embedding layer 
# input vocab defines the size of the vocabulary: the maximum number of words in the vocabulary
# output_dim defines the size of the output vector or the dimensions of the vector.
model.add(layers.Embedding(input_dim=1000, output_dim=64))

# To add LSTM layer: use layers.LSTM and to define number of units pass an integer.
model.add(layers.LSTM(128))

# layers.Dense(x) adds dense layer with x units.
model.add(layers.Dense(10))
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, LSTM, Dense, Dropout
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix
import plotly.io as pio
pio.renderers.default = "svg"

print("TensorFlow version:", tf.__version__)
```

## Part 1: Load and Prepare Sentiment Data (20 points)
In this lab, you will use a small sentiment dataset. You may either reuse the Twitter dataset from Week 3 or work with the provided example below.

```python
# Example Dataset
# (Optional: Uncomment and adapt the code if using train.csv from Week 3.)
texts = [
    "I love this movie, it's absolutely fantastic!",
    "This was the worst film I've ever seen.",
    "Great acting and brilliant direction.",
    "Terrible plot, boring characters, waste of time.",
    "An okay movie, nothing special.",
    "Incredible visuals and emotional depth.",
    "Predictable and poorly written.",
    "A masterpiece of modern cinema!"
] * 100

labels = [1, 0, 1, 0, 1, 1, 0, 1] * 100  # 1 = positive, 0 = negative

print(f"Loaded {len(texts)} samples")
```

```python
# Text Preprocessing
max_words = 1000
max_len = 50

tokenizer = Tokenizer(num_words=max_words, oov_token="<OOV>")
tokenizer.fit_on_texts(texts)

sequences = tokenizer.texts_to_sequences(texts)
padded = pad_sequences(sequences, maxlen=max_len, padding='post')

X_train, X_test, y_train, y_test = train_test_split(
    padded, labels, test_size=0.2, random_state=42
)

print("Data ready!")
```

## Part 2: Build and Train Models (40 points)

Tasks (Answer in Markdown Cells)

    1. Which model converged faster? Which achieved higher validation accuracy?

```python
# Model 1 Simple RNN

# Ensure arrays so validation_split works
X_train = np.array(X_train)
y_train = np.array(y_train)

# Model 1 Simple RNN
rnn_model = Sequential([
    Embedding(input_dim=max_words, output_dim=64, input_length=max_len),
    SimpleRNN(32),
    Dense(16, activation='relu'),
    Dropout(0.5),
    Dense(1, activation='sigmoid')
])

rnn_model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

rnn_model.summary()

history_rnn = rnn_model.fit(
    X_train, y_train,
    epochs=20,
    validation_split=0.2,
    verbose=1
)
```

```python
# Model 2 LSTM
lstm_model = Sequential([
    Embedding(max_words, 64, input_length=max_len),
    LSTM(32),
    Dense(16, activation='relu'),
    Dropout(0.5),
    Dense(1, activation='sigmoid')
])

lstm_model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

lstm_model.summary()

history_lstm = lstm_model.fit(
    X_train, y_train,
    epochs=10,
    validation_split=0.2,
    verbose=1
)
```

```python
# Evaluate Both Models
def plot_history(hist, title):
    plt.figure(figsize=(10,4))

    plt.subplot(1,2,1)
    plt.plot(hist.history['accuracy'], label='Train Acc')
    plt.plot(hist.history['val_accuracy'], label='Val Acc')
    plt.title(f'{title} - Accuracy')
    plt.legend()

    plt.subplot(1,2,2)
    plt.plot(hist.history['loss'], label='Train Loss')
    plt.plot(hist.history['val_loss'], label='Val Loss')
    plt.title(f'{title} - Loss')
    plt.legend()

    plt.show()

plot_history(history_rnn, "Simple RNN")
plot_history(history_lstm, "LSTM")
```
